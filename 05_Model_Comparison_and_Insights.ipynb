### Model Comparison and Business Insights

This notebook compares the performance of different models used in this major project:
- Linear Regression
- ARIMA
- LSTM (Deep Learning)

The goal is to identify the best-performing model and derive meaningful business insights.

```python
import pandas as pd
import numpy as np
```

```python
# Create a comparison table using results from previous notebooks

results = {
    "Model": ["Linear Regression", "ARIMA", "LSTM"],
    "MAE": [9.942187640773376, 6.482716236875971, 36.728427785503925],
    "RMSE": [12.278169391214652, 9.740675584248677, 45.34631286208073],
    "Remarks": [
        "Baseline ML model",
        "Classical time-series model",
        "Deep learning model (best)"
    ]
}

comparison_df = pd.DataFrame(results)
comparison_df
```

```python
best_model = comparison_df.loc[comparison_df["RMSE"].idxmin()]
best_model
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,4))
plt.bar(comparison_df["Model"], comparison_df["RMSE"])
plt.xlabel("Model")
plt.ylabel("RMSE")
plt.title("Model Comparison Based on RMSE")
plt.show()
```

## Model Interpretation

- **Linear Regression** acts as a baseline model and performs reasonably well, but it is limited in capturing complex temporal patterns.
- **ARIMA** improves performance by modeling time-series dependencies and stationarity.
- **LSTM** outperforms other models due to its ability to learn long-term dependencies in sequential data.

Based on RMSE values, **LSTM is the best-performing model** for Netflix stock price prediction.


## Business Insights

1. Netflix stock price is highly dependent on its historical prices, indicating strong temporal correlation.
2. Energy market variables such as crude oil and natural gas prices show indirect influence on stock price movement.
3. Short-term forecasting (7–30 days) is more reliable than long-term forecasting.
4. Deep learning models like LSTM provide better accuracy for volatile financial data.

These insights can help investors and analysts make informed decisions.


## Limitations
